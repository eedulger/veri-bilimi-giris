```python
vize = float(input("Vize Notunuzu Giriniz:"))
final = float(input("Final Notunuzu Giriniz:"))

ortalama = (vize*0.4) + (final*0.6)

print(f"Ortalama, {ortalama}")

if ortalama >= 50:
    print("Geçtiniz")
else:
    print("Dersten Kaldınız")

if ortalama >= 90:
    print("AA")
elif ortalama >= 80:
    print("BA")
elif ortalama >= 70:
    print("BB")
elif ortalama >= 60:
    print("CB")
elif ortalama >= 50:
    print("CC")
else:
    print("FF")
```
