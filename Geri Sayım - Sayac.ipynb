```python
import time

print("Geri Sayım - Sayaç")

dakika = int(input("Kaç dakika?: "))
saniye = int(input("Kaç saniye?: "))

toplam_saniye = (dakika * 60) + saniye

print("\nSüre Başladı\n")

while toplam_saniye >= 0:
    dk = toplam_saniye // 60
    sn = toplam_saniye % 60
    print(f"\rKalan Süre: {dk:02d} dk {sn:02d} sn", end="", flush = True)
    
    time.sleep(1)
    toplam_saniye -= 1

print("\rSüre Bitti!")

    
```
