```python
import random

print("Sayı Tahmin Oyunu")

alt_sinir = int(input("Aralığın alt sınırı: ")) 
ust_sinir = int(input("Aralığın üst sınırı: "))

tutulan_sayi = random.randint(alt_sinir, ust_sinir)

print(f"\nTamamdır. {alt_sinir} ile {ust_sinir} arasında bir sayı tuttum. Bakalım bulabilecek misin?")

while True:
    tahmin = int(input("Tahminini Gir: "))
    if tutulan_sayi > tahmin:
        print("Daha büyük bir sayı söylemelisin.")
    elif tutulan_sayi < tahmin:
        print("Daha küçük bir sayı söylemelisin.")
    else:
        print(f"TEBRİKLER! Doğru bildin. Tuttuğum sayı: {tutulan_sayi}")
        break
```
