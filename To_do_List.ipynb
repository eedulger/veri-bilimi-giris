```python
gorevler = []

print("TO_DO LİST")

while True:
    print("\nNe yapmak istiyorsun?")
    print("1- Görev Ekle")
    print("2- Görevleri Göster")
    print("3- Görev Sil")
    print("4- Çıkış Yap")
    
    # Bu kısımların hepsi while döngüsünün içine (içeriye doğru boşluk bırakılarak) taşındı:
    secim = input("Seçimin (1, 2, 3, 4): ")

    if secim == "1":
        yeni_gorev = input("Hangi görevi eklemek istiyorsun? : ")
        gorevler.append(yeni_gorev)
        print("Görev Eklendi")

    elif secim == "2":
        print("\n Görevlerin")
        if len(gorevler) == 0:
            print("Listede görev yok")
        else:
            print(gorevler)
            
    elif secim == "3":
        silinecek = input("Silinecek görev: ")
        if silinecek in gorevler:
            gorevler.remove(silinecek)
            print("Görev Silindi")
        else:
            print("Böyle bir görev bulunamadı")
            
    elif secim == "4":
        print("Uygulamadan çıkış yapılıyor")
        break
        
    else:
        print("Geçersiz bir seçim yaptın, tekrar dene.")
```
